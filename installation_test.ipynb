## 5. Create Data Programmatically

This example creates a small dataset directly in Python.

No external files are required.

The goal is to demonstrate how Python can create and manipulate tabular data using pandas.

````python
import pandas as pd
import numpy as np

np.random.seed(42)

example_data = pd.DataFrame({
    "sample_id": [f"sample_{i}" for i in range(1, 11)],
    "measurement": np.random.normal(50, 10, 10)
})

example_data
````

## 6. Create a Simple Plot

This cell creates a simple plot using the data generated above.

If the figure appears below the cell, matplotlib is working correctly.

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(
    example_data["sample_id"],
    example_data["measurement"],
    marker="o"
)

ax.set_title("Example Plot")
ax.set_xlabel("Sample")
ax.set_ylabel("Measurement")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
````
